```python
from pathlib import Path
import pandas as pd



csv_file = "/lustre/scratch127/cellgen/cellgeni/tickets/tic-4640/actions/palom_registration/all_paths.csv"
template_config = "conf_files/conf_BK22-SKI-27-FO-1-S34-D1.yaml"
generated_configs_dir = "conf_files"

# Base paths taken from your example
image_base_dir = Path("/lustre/scratch127/cellgen/cellgeni/tickets/tic-4640/data/AkoyaPhenocycler/xenium_aligned_palom")
output_base_dir = Path("/lustre/scratch127/cellgen/cellgeni/tickets/tic-4640/data/AkoyaPhenocycler/AKOYA_pipeline_out")


df = pd.read_csv(csv_file)
print(df)
if "Name" not in df.columns:
    raise ValueError("Column 'Name' not found in the CSV file.")

with open(template_config, "r") as f:
    template_lines = f.readlines()

# Create output directory for generated YAML files
Path(generated_configs_dir).mkdir(parents=True, exist_ok=True)

```

```python
# =========================
# GENERATE CONFIGS
# =========================

for samplename in df["Name"].dropna().astype(str):
    samplename = samplename.strip()

    new_image_path = image_base_dir / f"{samplename}_moving_reg.ome.tif"
    new_output_dir = output_base_dir / samplename
    new_config_name = f"conf_{samplename}.yaml"
    new_config_path = Path(generated_configs_dir) / new_config_name

    new_lines = []
    for line in template_lines:
        if line.startswith("image_path:"):
            new_lines.append(f"image_path: {new_image_path}\n")
        elif line.startswith("output_dir:"):
            new_lines.append(f"output_dir: {new_output_dir}\n")
        else:
            new_lines.append(line)

    with open(new_config_path, "w") as f:
        f.writelines(new_lines)

    print(f"Created: {new_config_path}")
```
